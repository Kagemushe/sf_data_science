#### <center>Равномерное распределение<center>

```python
import numpy as np
from scipy.stats import randint
```

```python
# Задаем возможные значения случайной величины
x = np.arange(1,7)
# Задаем интервал, на котором будут распределятся вероятности
disc_uni_dist = randint(1,7)
# Вычисляем вероятности выпадения каждого значения случайной величины
pmf = disc_uni_dist.pmf(x)
pmf
```

```python
# Рассчитаем кумулятивную вероятность
cdf = disc_uni_dist.cdf(x)
print(cdf)
```

```python
# Задаем возможные значения случайной велечины
x = np.arange(1,251)
# Задаем интервал, на котором будут распределятся наши вероятности
disc_uni_dist = randint(1,251)
# Вычисляем вероятности выпадения каждого значения случайной величины
cdf = disc_uni_dist.cdf(x)
# Ищем вероятность для кото-то из первых пятидесяти участников
print(cdf[49])
```

#### <center>Расспределение Бернулли<center>

```python
from scipy.stats import bernoulli
import seaborn as sns

data = bernoulli.rvs(size=500, p=0.6)
ax = sns.displot(data, kde=False, color='dodgerblue')
ax.set(xlabel='Значение случайной величины', ylabel='Частота')
```

```python
unique, counts = np.unique(data, return_counts=True)
print(np.asarray((unique, counts)).T)
```

#### <center>Биномиальное распределение<center>

```python
import scipy

scipy.stats.binom.pmf(8,10,0.8)+scipy.stats.binom.pmf(9,10,0.8)+scipy.stats.binom.pmf(10,10,0.8)
```

```python
from numpy import random
import seaborn as sns

data = random.binomial(n=10, p=0.5, size=1000)
ax = sns.displot(data, kde=False, color='dodgerblue')
ax.set(xlabel='Значение случайной величины', ylabel='Частота')
```

#### <center>Распределение Пуасонна<center>

```python
import scipy
print(scipy.stats.distributions.poisson(5, 4.5))
```

```python
from numpy import random
import seaborn as sns

data = random.poisson(lam=3, size=1000)
ax = sns.displot(data, kde=False, color='dodgerblue')
ax.set(xlabel='Значение случайной величины', ylabel='Частота')
```
